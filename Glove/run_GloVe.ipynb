# GloVe Word Embeddings Pipeline

This notebook runs the complete GloVe (Global Vectors for Word Representation) pipeline for sentiment classification on Twitter data.

## Pipeline Overview

1. **Build Vocabulary**: Extract all words from training data
2. **Cut Vocabulary**: Filter words appearing less than 5 times
3. **Pickle Vocabulary**: Serialize vocabulary for fast loading
4. **Build Co-occurrence Matrix**: Compute word co-occurrence statistics
5. **Train GloVe Embeddings**: Learn word embeddings using matrix factorization
6. **Build Features**: Convert tweets to feature vectors using embeddings
7. **Train Classifier**: Train MLP classifier on extracted features
8. **Predict Test Set**: Generate predictions for test data


```python
import os
import subprocess
import sys

# Set up paths - notebook should be run from Glove directory
SCRIPT_DIR = os.getcwd()
if not SCRIPT_DIR.endswith('Glove'):
    # If running from parent directory, change to Glove
    if os.path.exists('Glove'):
        os.chdir('Glove')
        SCRIPT_DIR = os.getcwd()
    else:
        # Try to find Glove directory
        parent = os.path.dirname(SCRIPT_DIR)
        glove_path = os.path.join(parent, 'Glove')
        if os.path.exists(glove_path):
            os.chdir(glove_path)
            SCRIPT_DIR = os.getcwd()

print(f"Working directory: {os.getcwd()}")
print(f"Script directory: {SCRIPT_DIR}")
```

## Step 1: Build Vocabulary

Extract all words from the training data and count their occurrences.


```python
# Run build_vocab.sh
subprocess.run(['bash', 'build_vocab.sh'], capture_output=True, text=True, cwd=SCRIPT_DIR)
print("✓ Vocabulary built successfully")
```

## Step 2: Cut Vocabulary

Filter vocabulary to keep only words appearing at least 5 times.


```python
# Run cut_vocab.sh
subprocess.run(['bash', 'cut_vocab.sh'], capture_output=True, text=True, cwd=SCRIPT_DIR)
print(f"✓ Vocabulary cut completed")
```

## Step 3: Pickle Vocabulary

Serialize the vocabulary dictionary for fast loading in subsequent steps.


```python
# Run pickle_vocab.py
subprocess.run([sys.executable, 'pickle_vocab.py'], capture_output=True, text=True, cwd=SCRIPT_DIR)
print(f"✓ Vocabulary pickled successfully.")
```

## Step 4: Build Co-occurrence Matrix

Compute word co-occurrence statistics from the training corpus. This step may take several minutes.


```python
# Run cooc.py
subprocess.run([sys.executable, 'cooc.py'], cwd=SCRIPT_DIR)
print(f"✓ Co-occurrence matrix built successfully")

```

## Step 5: Train GloVe Embeddings

Learn word embeddings using stochastic gradient descent on the co-occurrence matrix. This is the core training step.


```python
# Run glove_solution_mini_batch.py
result = subprocess.run([sys.executable, 'glove_solution_mini_batch.py'], cwd=SCRIPT_DIR)
print(f"✓ GloVe embeddings trained successfully")
```

## Step 6: Build Features

Convert training tweets into feature vectors by averaging word embeddings. This step processes all training data.


```python
# Run build_features.py
result = subprocess.run([sys.executable, 'build_features.py'], cwd=SCRIPT_DIR)
print(f"✓ Features built successfully")
```

## Step 7: Train Classifier

Train a Multi-Layer Perceptron (MLP) classifier on the extracted features. The model will be tuned using validation set.


```python
# Run train_classifier.py
subprocess.run([sys.executable, 'train_classifier.py'], cwd=SCRIPT_DIR)
print("✓ Classifier trained successfully")
```

## Step 8: Predict Test Set

Generate predictions for the test data and create submission file.


```python
# Run predict_test.py
subprocess.run([sys.executable, 'predict_test.py'], cwd=SCRIPT_DIR)
print(f"✓ Predictions generated successfully")
```

## Pipeline Complete! 🎉

All steps have been executed successfully. The submission file `submission_glove.csv` is ready for evaluation.

Upon submission to AIcrowd, `submission_glove.csv` yielded an accuracy of 82.7% on the test set.
