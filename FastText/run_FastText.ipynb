# FastText Word Embeddings Pipeline

This notebook runs the complete FastText pipeline for sentiment classification on Twitter data.

## Pipeline Overview

1. **Build Training Corpus**: Merge positive and negative training data into a single file
2. **Train FastText Model**: Learn word embeddings using unsupervised learning (skipgram/CBOW)
3. **Build Features**: Convert tweets to feature vectors using FastText embeddings
4. **Train Classifier**: Train MLP classifier on extracted features
5. **Predict Test Set**: Generate predictions for test data


```python
# ! pip install fasttext
```

```python
import os
import subprocess
import sys

# Set up paths - notebook should be run from FastText directory
SCRIPT_DIR = os.getcwd()
if not SCRIPT_DIR.endswith('FastText'):
    # If running from parent directory, change to FastText
    if os.path.exists('FastText'):
        os.chdir('FastText')
        SCRIPT_DIR = os.getcwd()
    else:
        # Try to find FastText directory
        parent = os.path.dirname(SCRIPT_DIR)
        fasttext_path = os.path.join(parent, 'FastText')
        if os.path.exists(fasttext_path):
            os.chdir(fasttext_path)
            SCRIPT_DIR = os.getcwd()

print(f"Working directory: {os.getcwd()}")
print(f"Script directory: {SCRIPT_DIR}")
```

## Step 1: Build Training Corpus

Merge positive and negative training data into a single file for unsupervised FastText training.


```python
# Run build_all_train_txt.py
subprocess.run([sys.executable, 'build_all_train_txt.py'], cwd=SCRIPT_DIR)
print(f"✓ Training corpus built successfully")
```

## Step 2: Train FastText Model

Train unsupervised FastText embeddings using skipgram or CBOW. This step learns word representations that capture subword information.


```python
# Run train_fasttext_unsup.py
subprocess.run([sys.executable, 'train_fasttext_unsup.py'], cwd=SCRIPT_DIR)
print(f"✓ FastText model trained successfully")
```

## Step 3: Build Features

Convert training tweets into feature vectors using FastText embeddings. This step processes all training data.


```python
# Run build_features_fasttext.py
subprocess.run([sys.executable, 'build_features_fasttext.py'], cwd=SCRIPT_DIR)
print(f"✓ Features built successfully")
```

## Step 4: Train Classifier

Train a Multi-Layer Perceptron (MLP) classifier on the extracted features. The model will be tuned using validation set.


```python
# Run train_mlp_fasttext.py
subprocess.run([sys.executable, 'train_mlp_fasttext.py'], cwd=SCRIPT_DIR)
print(f"✓ Classifier trained successfully")
```

## Step 5: Predict Test Set

Generate predictions for the test data and create submission file.


```python
# Run predict_fasttext.py
subprocess.run([sys.executable, 'predict_fasttext.py'], cwd=SCRIPT_DIR)
print(f"✓ Predictions generated successfully")
```

## Pipeline Complete! 🎉

All steps have been executed successfully. The submission file `submission_fasttext.csv` is ready for evaluation.

### Summary

- **Training Corpus**: Merged positive and negative tweets into single file
- **FastText Model**: Trained unsupervised word embeddings with subword information
- **Features**: Extracted using mean and max pooling of FastText embeddings
- **Classifier**: Trained MLP model with best hyperparameters
- **Predictions**: Generated for test set

You can now submit `submission_fasttext.csv` to the evaluation platform.
